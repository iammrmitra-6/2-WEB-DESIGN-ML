### 1. The `Path()` Function

The `Path()` function in FastAPI provides metadata, validation rules, and documentation hints specifically for **path parameters** inside API endpoints.

```python
from fastapi import FastAPI, Path

app = FastAPI()

@app.get("/items/{item_id}")
async def read_item(
    item_id: int = Path(..., title="The ID of the item", ge=1, le=1000)
):
    return {"item_id": item_id}

```

#### Validation Parameters Explanation

* **`title`**: Sets a short human-readable title for documentation generators like Swagger UI.
* **`description`**: Provides a longer explanation of what the path parameter represents.
* **`example`**: Provides a sample value displayed in the automatic interactive documentation.
* **`ge` (Greater than or equal to)**: Ensures the parameter value is greater than or equal to the specified number (e.g., `ge=1` enforces $item\_id \ge 1$).
* **`gt` (Greater than)**: Ensures the parameter value is strictly greater than the specified number (e.g., `gt=0` enforces $item\_id > 0$).
* **`le` (Less than or equal to)**: Ensures the parameter value is less than or equal to the specified number (e.g., `le=1000` enforces $item\_id \le 1000$).
* **`lt` (Less than)**: Ensures the parameter value is strictly less than the specified number (e.g., `lt=100` enforces $item\_id < 100$).
* **`min_length`**: Enforces a minimum character length for string parameters.
* **`max_length`**: Enforces a maximum character length for string parameters.
* **`regex` / `pattern**`: Requires the string parameter to match a specific regular expression pattern.

### 2. HTTP Status Codes

HTTP status codes are 3-digit numbers returned by a web server to inform the client (browser, frontend framework, or mobile app) about the outcome of an HTTP request.

```
+--------+       Request (e.g., GET /patients)       +--------+
| Client | ----------------------------------------> | Server |
| (App)  | <---------------------------------------- |(FastAPI|
+--------+      Response (e.g., 200 OK + JSON)       +--------+

```

#### Status Code Categories

* **2xx Series (Success)**: Indicates that the client's request was successfully received, understood, and accepted by the server.
* **3xx Series (Redirection)**: Indicates that further action must be taken by the client to complete the request, such as following a redirect link.
* **4xx Series (Client Error)**: Indicates that the request contains bad syntax or invalid data, meaning something went wrong on the client side.
* **5xx Series (Server Error)**: Indicates that the server failed to fulfill an apparently valid request due to an issue on the backend.

#### Common HTTP Status Codes Explanation

* **200 OK**: The standard response for successful HTTP requests. Commonly returned when a GET request fetches data or a POST request processes data successfully without creating a new resource.
* **201 Created**: Indicates that the request succeeded and a new resource was created as a result. Commonly returned after a POST request that adds a new record to a database.
* **204 No Content**: Indicates that the server successfully processed the request, but there is no payload to return. Commonly used after a DELETE operation.
* **400 Bad Request**: Returned when the server cannot process the request due to client error, such as malformed JSON, a missing required payload field, or an incorrect data type.
* **401 Unauthorized**: Indicates that authentication is required and has failed or has not yet been provided. The client must log in or provide a valid access token.
* **403 Forbidden**: Returned when the client is authenticated, but lacks the necessary permissions or rights to access the requested resource.
* **404 Not Found**: Indicates that the requested resource could not be found on the server, such as searching for a patient ID that does not exist in the database.
* **500 Internal Server Error**: A generic error message returned when an unexpected condition or unhandled exception occurs on the backend server.
* **502 Bad Gateway**: Returned when a proxy or gateway server (like Nginx) receives an invalid response from the backend application server (like FastAPI/Uvicorn).
* **503 Service Unavailable**: Indicates that the server is currently unable to handle the request due to a temporary overload or scheduled maintenance.

### 3. FastAPI `HTTPException`

`HTTPException` is a built-in exception class in FastAPI used to return custom HTTP error responses when errors occur during execution, preventing server crashes and delivering structured JSON error responses.

```python
from fastapi import FastAPI, HTTPException, status

app = FastAPI()

patients_db = {1: "Alice"}

@app.get("/patients/{patient_id}")
async def get_patient(patient_id: int):
    if patient_id not in patients_db:
        raise HTTPException(
            status_code=status.HTTP_404_NOT_FOUND,
            detail="Patient ID not found in database",
            headers={"X-Error-Reason": "Invalid ID"}
        )
    return {"patient": patients_db[patient_id]}

```

#### Key Arguments

* **`status_code`**: The HTTP status code number to be returned to the client (e.g., 404, 400, 403).
* **`detail`**: A string or complex structured object containing human-readable error details sent in the JSON response body.
* **`headers`**: An optional dictionary of custom HTTP response headers to send back with the error response.

### 4. Query Parameters & The `Query()` Function

#### Query Parameters Explained

Query parameters are optional key-value pairs appended to the end of a URL after a `?` symbol. They allow clients to send additional instructions to the backend for operations like filtering, sorting, searching, and pagination, without modifying the base URL path itself.

```
URL: /patients?city=Delhi&sort_by=age
             ^ ^~~~~~~~~~ ^~~~~~~~~~~~
             |     |           |
    Start Marker  Key-Value 1  Key-Value 2 (separated by &)

```

* **The `?` symbol**: Marks the beginning of the query string in a URL.
* **The `&` symbol**: Used to separate multiple key-value pairs in a single URL.
* **Filtering example (`city=Delhi`)**: Restricts the returned dataset to patients living in Delhi.
* **Sorting example (`sort_by=age`)**: Orders the filtered results based on the age field.

#### The `Query()` Function

`Query()` is a utility function provided by FastAPI to declare metadata, validation constraints, and default values for query parameters in endpoint parameters.

```python
from typing import Optional
from fastapi import FastAPI, Query

app = FastAPI()

@app.get("/patients")
async def read_patients(
    city: Optional[str] = Query(None, max_length=50, description="Filter by city"),
    sort_by: str = Query("age", regex="^(age|name|created_at)$")
):
    return {"city": city, "sort_by": sort_by}

```
